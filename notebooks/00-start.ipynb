# 강의 개요 — 이 코스로 무엇을, 어떻게 배우나

> ⏱ 15분 · 읽기 전용

이 코스는 **AI를 직접 이해하고 만들 수 있는 사람**을 목표로 합니다. 머신러닝의 첫 모델에서 시작해 딥러닝의 역전파를 손으로 짜고, GPT를 밑바닥부터 만들어 사전학습하고, 실제 LLM을 파인튜닝·평가하고, 이미지와 언어를 잇는 VLM까지 직접 학습시킵니다. 모든 코드는 실제로 실행해 검증했으며, 무료 GPU(Google Colab)에서 돌아갑니다.

## 이 코스를 마치면 할 수 있는 것 (학습 성과)

1. 데이터를 나누고, 기준선을 세우고, 검증 데이터로 모델을 고르고, 테스트 데이터로 정직하게 보고할 수 있다.
2. 신경망의 순전파·역전파를 수식 없이 코드로 설명하고, `loss.backward()`가 하는 일을 numpy로 재현할 수 있다.
3. PyTorch로 모델 구조를 정의·수정·저장하고, loss 곡선을 읽어 과적합·학습률 문제를 진단하고 고칠 수 있다.
4. 트랜스포머(어텐션, 잔차, LayerNorm)를 직접 구현해 언어모델을 사전학습하고, BPE·혼합 정밀도·스케줄러 같은 실전 요소를 붙일 수 있다.
5. 사전학습된 LLM을 프롬프트·RAG·LoRA SFT·DPO로 원하는 행동에 맞추고, 그 효과와 부작용을 **숫자로** 평가할 수 있다.
6. 이미지 인코더와 LLM을 이어 VLM을 만들고, 실제 VLM을 내 이미지 과제에 파인튜닝할 수 있다.
7. 새 문제를 받았을 때 프롬프트 → RAG → 파인튜닝 순으로 접근을 고르고, GPU 메모리와 데이터 양을 어림해 계획을 세울 수 있다.
8. 모델의 한계(환각, 편향, 망각)와 데이터·배포의 책임을 알고 프로젝트 점검표를 적용할 수 있다.

## 12주 학습 계획 (권장)

주당 6~8시간 기준입니다. 빠르면 6주에, 천천히 하면 한 학기에 맞춥니다.

| 주 | 레슨 | 이 주의 목표 |
|---|---|---|
| 1 | 0부 전체, 첫 모델 | 환경을 갖추고 첫 모델의 정확도를 낸다 |
| 2 | 과적합과 검증, 표 데이터와 트리 | train/val/test의 차이를 몸으로 안다 |

| 4 | 역전파 직접 구현, CNN | `backward()`를 직접 짜 본다. 모델 구조를 바꿔 본다 |
| 5 | 학습 잘 시키는 법, 전이학습 | loss 곡선을 보고 진단한다. 사전학습 모델을 쓴다 |

## 평가 방식 (스스로 채점)

이 코스에는 시험이 없습니다. 대신 레슨마다 세 가지 장치가 있고, 마지막에 프로젝트가 있습니다.

| 장치 | 무엇을 확인하나 | 통과 기준 |
|---|---|---|
| **스스로 점검** (각 레슨) | 개념을 말로 설명할 수 있는가 | 답을 보기 전에 3문항 중 2문항 이상 설명 |
| **직접 고쳐보기** (각 레슨) | 코드를 바꿔 결과를 예측·확인할 수 있는가 | 번호 없는 문항 전부 + 도전 1개 |
| **핵심 정리** (각 레슨) | 다음 레슨으로 갈 준비가 됐는가 | 다섯 줄을 보지 않고 쓸 수 있음 |
| **종합 프로젝트** (5부) | 처음부터 끝까지 혼자 만들 수 있는가 | 채점 기준표 70점 이상 |

각 실습 문항에는 접힌 **힌트와 예상 결과**가 있습니다. 먼저 스스로 예측하고 실행한 뒤에 펼쳐 보세요. 예상과 다를 때가 가장 많이 배우는 순간입니다.

## 모든 레슨을 관통하는 네 줄

[필요한 수학](https://yun-sooyong.github.io/dl-study/#02-math-minimum)에서 개념으로, [텐서와 자동미분](https://yun-sooyong.github.io/dl-study/#20-tensor-autograd)에서 코드로 만나는 아래 네 줄이 마지막 VLM 학습에서도 **그대로** 나옵니다. 모델이 커지고 데이터가 바뀔 뿐입니다.

```
예측 = 모델(입력)            # forward
손실 = 손실함수(예측, 정답)     # 얼마나 틀렸나
손실.backward()              # 각 파라미터가 손실에 미친 영향(기울기) 계산
옵티마이저.step()             # 기울기 반대 방향으로 파라미터를 조금 이동
```

## 레슨의 구성

모든 레슨은 같은 순서입니다.

1. **목표** — 이 레슨이 끝나면 할 수 있게 되는 것
2. **본문과 코드** — 설명을 읽고, 실행하고, 출력을 확인. 코드 블록마다 **코드 읽기**가 따라붙어 "왜 이 함수·이 값인가"를 설명합니다
3. **핵심 정리** — 기억할 것 다섯 줄
4. **스스로 점검** — 답을 가린 퀴즈. 막히면 그 부분을 다시 읽으라는 신호입니다
5. **직접 고쳐보기** — 코드를 바꿔 보는 실습. **진짜 수업은 여기**입니다
