# 0. 시작하기 — 로드맵과 실습 환경

이 코스의 목표는 하나입니다. **모델을 직접 고치고, 직접 만들고, 직접 학습시켜 보는 것.**
이론을 다 알고 시작할 필요는 없습니다. 코드를 돌려보고, 숫자를 바꿔보고, 망가뜨려 보면서 배웁니다.

## 로드맵

| 단계 | 만드는 것 | 배우는 것 |
|---|---|---|
| 1 | 직선 하나 맞추기 | 텐서, 자동미분, 경사하강법 — 학습의 본질 |
| 2 | 손글씨 숫자 분류기 (MLP) | `nn.Module`, DataLoader, **학습 루프** |
| 3 | 모델 구조 바꾸기 (CNN) | 모델을 **수정**하는 법, 저장/불러오기 |
| 4 | 나만의 미니 GPT | 토큰, 어텐션, 트랜스포머 — LLM을 **밑바닥부터** |
| 5 | 진짜 LLM 파인튜닝 | Hugging Face, LoRA — 남이 만든 LLM을 **내 데이터로** |
| 6 | 나만의 미니 VLM | 이미지 인코더 + LLM을 **이어붙여** 학습 |
| 7 | 진짜 VLM 파인튜닝 | 사전학습된 VLM 사용과 LoRA 학습 |

모든 단계는 같은 뼈대를 공유합니다. 1단계에서 배우는 아래 네 줄이 7단계 VLM 학습에서도 그대로 나옵니다.

```
예측 = 모델(입력)          # forward
손실 = 손실함수(예측, 정답)   # 얼마나 틀렸나
손실.backward()            # 각 파라미터가 손실에 미친 영향(기울기) 계산
옵티마이저.step()           # 기울기 반대 방향으로 파라미터를 조금 이동
```

## 실습 환경: Google Colab

GPU가 없어도 됩니다. 각 레슨 상단의 **▶ Colab에서 실행하기** 버튼을 누르면 무료 GPU가 달린 노트북이 열립니다.

1. 버튼을 눌러 Colab을 연다 (구글 계정 필요)
2. 메뉴 **런타임 → 런타임 유형 변경 → T4 GPU** 선택 (1~3단계는 CPU로도 충분)
3. 셀을 위에서부터 `Shift+Enter`로 하나씩 실행
4. 내 수정본을 남기려면 **파일 → Drive에 사본 저장**

> 셀을 그냥 실행만 하고 넘어가지 마세요. 각 레슨 끝의 **직접 고쳐보기**가 진짜 수업입니다.
